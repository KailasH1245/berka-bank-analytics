```python
import urllib
import pandas as pd
from sqlalchemy import create_engine

SERVER_NAME = r'LAPTOP-67Q06DO1\SQLEXPRESS02'
DATABASE_NAME = 'BerkaBankDB'
CLEAN_DIR = 'cleaned-data'

params = urllib.parse.quote_plus(
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={SERVER_NAME};"
    f"DATABASE={DATABASE_NAME};"
    f"Trusted_Connection=yes;"
)

engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}", fast_executemany=True)

table_names = ['district', 'client', 'account', 'disp', 'card', 'loan', 'order', 'trans']

for name in table_names:
    df = pd.read_csv(f"{CLEAN_DIR}/{name}_clean.csv", low_memory=False)
    df.to_sql(
        name=name,
        con=engine,
        if_exists='replace',
        index=False,
        chunksize=10000
    )
    print(f"Loaded {name:<10} -> {len(df):>8,} rows into SQL Server")
```
